```python
import sys

sys.path.append('../src/')
import numpy as np
import scipy.io as sio
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
import pysindy as ps
from libselection.utils import *
from libselection import *
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
import matplotlib as mpl
```

```python
data = sio.loadmat('../Data/kuramoto_sivishinky.mat')
print(data.keys())
u = data['uu'].real
x = data['x'][...,0]
t = data['tt'][0]
dt = t[1]-t[0]
dx = x[2]-x[1]

n = len(x)
m = len(t)
print(f'Shape of u: {u.shape}, shape of x: {x.shape}, shape of t: {t.shape}')
```

```python
X, T = np.meshgrid(x, t)
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, T, u.T, rstride=1, cstride=1, cmap=plt.colormaps.get('coolwarm'),
    linewidth=0, antialiased=False)
# plt.title('Burgers Equation', fontsize = 20)

ax.set_axis_off()
ax.set_aspect('equal')
fig.tight_layout()
ax.view_init(elev=30, azim=-80)
fig.savefig('Figs/ks_equation_example.png', dpi=1200, format='png', bbox_inches='tight',pad_inches=0,)
```
